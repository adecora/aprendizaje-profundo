```python
#| include: false
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import tensorflow as tf
from matplotlib.ticker import FuncFormatter, MultipleLocator


plt.style.use('ggplot')
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"] = 100
```

## Implementación de dos redes de perceptrones multicapa (MLP)

En la figura @fig-sin-x se muestra el seno de $x$ en el intervalo de $0$ a $2\pi$.

```python
#| echo: false
#| label: fig-sin-x
#| fig-cap: "Seno de $x$"
#| fig-align: center
x = np.linspace(0, np.deg2rad(360), 100)
y = np.sin(x)

fig, ax = plt.subplots()

ax.plot(x, y, color='red', label='$sin(x)$')
ax.legend()

ax.set_xlabel('$x$')
ax.set_ylabel('$sin(x)$')

ax.xaxis.set_major_locator(MultipleLocator(base=np.pi/2))
ax.xaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: rf'${x/np.pi:g}\pi$' if x != 0 else '$0$')
)

# ax.grid()
# ax.spines[['top', 'right']].set_visible(False)

plt.show()
```
